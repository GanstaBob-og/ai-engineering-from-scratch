```python
arrs = []
for _ in range(5):
    arr = np.random.randn(50) * .1
    arrs.append(arr)

arrs = np.array(arrs, dtype=float)

i, j = np.triu_indices(5, k=1)  # уникальные пары i < j
pairs = np.stack([arrs[i], arrs[j]], axis=1)  # (10, 2, 50)

cos = []
for pair in pairs:
    a, b = pair[0], pair[1]
    cos.append((pair, np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))))

res = max(cos, key=lambda c: c[1])

print(f"Векторы \n{res[0][0]} и \n{res[0][1]}\n имеют максимальное значение схожести равное {res[1]}")
```
